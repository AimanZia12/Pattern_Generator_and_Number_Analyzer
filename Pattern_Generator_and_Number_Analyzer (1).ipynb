```python
print("\n-----Welcome to the Pattern Generator and Number Analyzer------")
while True:
    print("\n You have three choices.")
    print("1.Pattern Generator")
    print("2.Number Analyzer")
    print("3.Exit")
    
    user_choice = input("Enter your choice:")
    if user_choice == '1':
        print("\n------User selected the option of Pattern Generator:")
        print("\n Now you can generate three type of patterns.")
        print("1.Right-Angled Triangle")
        print("2.Pyramid")
        print("3.Left-Angled Triangle")
        
        pattern_generator_choice = input("Enter your pattern generator choice:")
        number_of_rows = int(input("Enter the number of rows:"))
        if number_of_rows <= 0:
            print("The number of rows are very less")
            continue
            
        if pattern_generator_choice == '1':
            print("\n User wanted to generate Right-Angled Triangle Pattern:")
            for i in range(1, number_of_rows + 1):
                for j in range(i):
                    print("*", end = " ")
                    pass
                print()
                
        elif pattern_generator_choice == '2':
            print("\n User wanted to generate Pyramid Pattern:")
            for i in range(number_of_rows):
                for j in range(number_of_rows - i - 1):
                    print(" ", end="")
                for k in range(2 * i + 1):
                        print("*", end="")
                print()
        elif pattern_generator_choice == '3':
            print("\n User wanted to generate Left-Angled Triangle Pattern:")
            for i in range(1, number_of_rows):
                for j in range(number_of_rows - i):
                    print(" ", end=" ")
                for k in range(i):
                    print("*", end=" ")
                print()
        else:
            print("No Pattern Selected")
            break
    elif user_choice == '2':
        print("\n------User selected the option of Number Analyzer:")
        Starting_Number = int(input("Enter your starting number:"))
        Ending_Number = int(input("Enter your ending number:"))
        if Ending_Number <= 0  or Ending_Number < Starting_Number:
            print("Ending Number cannot be negative or smaller than Starting Number")
            continue
        print("Checking if a number is even or odd")
        sum = 0
        for num in range(Starting_Number, Ending_Number+1):
            if num % 2 == 0:
                print(num,"is Even")
            else:
                print(num,"is Odd")
            sum += num
        print("The total sum of numbers is :", sum)
    else:
        print("\n------User selected the Exit Option----------")
        print("\n------Thank you, for using the Student Data Organizer------") 
        print("\n------Exiting the Program, Good Bye!---------")
        break



```
